# Corpus normalization inspection / بررسی فایل پیش‌بینی
Retrospective reconstruction; this is not an original timestamped development artifact and results require rerunning. Set `PREDICTIONS_JSON` to compute from local data.

```python
import json, os, re
from pathlib import Path
path = Path(os.environ['PREDICTIONS_JSON']) if os.environ.get('PREDICTIONS_JSON') else None
payload = json.loads(path.read_text(encoding='utf-8')) if path else []
rows = payload if isinstance(payload, list) else payload.get('records', payload.get('predictions', []))
{'selected': str(path) if path else None, 'records': len(rows)}
```

```python
def basic(text): return re.sub(r'\s+', ' ', re.sub(r'[,\.!?;:]', ' ', str(text).lower())).strip()
preview = []
for row in rows[:10]:
    value = row.get('hypothesis_raw', row.get('hypothesis', row.get('hyp', '')))
    preview.append({'before': value, 'after_basic': basic(value)})
preview
```
