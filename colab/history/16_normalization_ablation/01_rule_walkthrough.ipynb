# ATC normalization rules / قواعد نرمال‌سازی ATC
Retrospective reconstruction; this is not an original timestamped development artifact and results require rerunning. These labeled examples demonstrate transformations only.

```python
import re
digit = dict(zip('0123456789', 'zero one two three four five six seven eight nine'.split()))
def expand_digits(value): return ' '.join(digit[item] for item in value)
def normalize_demo(text):
    text = re.sub(r'\bniner\b', 'nine', text.lower())
    text = re.sub(r'\bfl\s*(\d{2,3})\b', lambda m: 'flight level '+expand_digits(m.group(1)), text)
    return re.sub(r'\s+', ' ', text).strip()
examples = ['descend FL180', 'niner thousand feet']
[(item, normalize_demo(item)) for item in examples]
```
