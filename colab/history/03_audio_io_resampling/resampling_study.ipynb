# Resampling study / مطالعه بازنمونه‌برداری

Tests the 16 kHz conversion assumption with in-band and out-of-band tones.

```python
import math
import numpy as np
from scipy.signal import resample_poly
source_rate, target_rate = 48000, 16000
t = np.arange(source_rate, dtype=np.float64)/source_rate
source = .4*np.sin(2*np.pi*1000*t) + .2*np.sin(2*np.pi*12000*t)
g = math.gcd(source_rate, target_rate)
converted = resample_poly(source, target_rate//g, source_rate//g)
len(source), len(converted)
```

The anti-aliasing filter should retain 1 kHz while suppressing the 12 kHz component above the target Nyquist limit.

```python
def tone_amplitude(x, rate, hz):
    n = np.arange(len(x)); basis = np.exp(-2j*np.pi*hz*n/rate)
    return float(2*np.abs(np.dot(x, basis))/len(x))
print({'1k_before': tone_amplitude(source, source_rate, 1000),
       '1k_after': tone_amplitude(converted, target_rate, 1000),
       'peak_after': float(np.max(np.abs(converted))), 'finite': bool(np.isfinite(converted).all())})
```
