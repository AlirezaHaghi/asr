# WAV I/O round trip / آزمون رفت‌وبرگشت WAV
Retrospective reconstruction created from the final notebook/report; results require rerunning.
A synthetic signal keeps this notebook executable without private recordings.

```python
from pathlib import Path
import numpy as np
import soundfile as sf
rate = 16000
t = np.arange(rate, dtype=np.float32) / rate
signal = 0.25*np.sin(2*np.pi*440*t) + 0.05*np.sin(2*np.pi*1200*t)
Path('artifacts').mkdir(exist_ok=True)
sf.write('roundtrip.wav', signal, rate, subtype='PCM_16')
loaded, loaded_rate = sf.read('roundtrip.wav', dtype='float32')
```

Quantization error is expected after PCM-16 encoding.

```python
error = loaded - signal
print({'rate': loaded_rate, 'frames': len(loaded), 'peak': float(np.max(np.abs(loaded))),
       'rmse_roundtrip': float(np.sqrt(np.mean(error**2))), 'all_finite': bool(np.isfinite(loaded).all())})
```
