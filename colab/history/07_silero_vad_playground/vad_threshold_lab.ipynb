# Silero threshold lab / آزمایش آستانه Silero
Retrospective reconstruction: this notebook defines a rerunnable sweep and does not contain claimed historical metrics.

```python
from pathlib import Path
import soundfile as sf
import torch
from scipy.signal import resample_poly
from silero_vad import load_silero_vad, get_speech_timestamps
AUDIO_PATH = Path('sample.wav')
THRESHOLDS = [0.25, 0.4, 0.5, 0.65, 0.8]
print({'audio_exists': AUDIO_PATH.exists(), 'thresholds': THRESHOLDS})
```

```python
def load_16k(path):
    x, sr = sf.read(path, dtype='float32', always_2d=True)
    x = x.mean(axis=1)
    return resample_poly(x, 16000, sr).astype('float32') if sr != 16000 else x

observations = []
if AUDIO_PATH.exists():
    x = load_16k(AUDIO_PATH)
    model = load_silero_vad()
    for threshold in THRESHOLDS:
        spans = get_speech_timestamps(torch.from_numpy(x), model, sampling_rate=16000, threshold=threshold)
        speech_s = sum(s['end'] - s['start'] for s in spans) / 16000
        observations.append({'threshold': threshold, 'segments': len(spans), 'speech_s': speech_s})
observations
```
