# VAD threshold sweep / بررسی آستانه VAD
Retrospective reconstruction; this is not an original timestamped development artifact and results require rerunning. This notebook demonstrates the measurement method on a labeled synthetic waveform.

```python
import numpy as np
sample_rate = 16000
rng = np.random.default_rng(7)
silence = rng.normal(0, 0.001, sample_rate).astype('float32')
speech_like = (0.08*np.sin(2*np.pi*180*np.arange(sample_rate)/sample_rate) + rng.normal(0, 0.006, sample_rate)).astype('float32')
audio = np.concatenate([silence, speech_like, silence])
len(audio) / sample_rate
```

```python
def active_ratio(signal, threshold_db, frame_ms=30):
    size = round(sample_rate * frame_ms / 1000)
    count = int(np.ceil(len(signal) / size))
    padded = np.pad(signal, (0, count*size-len(signal))).reshape(count, size)
    db = 20*np.log10(np.sqrt(np.mean(padded**2, axis=1)) + 1e-12)
    return float(np.mean(db >= threshold_db))
sweep = [{'threshold_db': t, 'active_frame_ratio': active_ratio(audio, t)} for t in (-50, -45, -40, -35, -30)]
sweep
```
