# Compare supplied VAD outputs / مقایسه خروجی‌های کاربر
Retrospective reconstruction; this is not an original timestamped development artifact and results require rerunning. Set `NO_VAD_JSON` and `VAD_JSON` to prediction files.

```python
import json
import os
from pathlib import Path
paths = {name: Path(os.environ[name]) for name in ('NO_VAD_JSON', 'VAD_JSON') if os.environ.get(name)}
{name: path.exists() for name, path in paths.items()}
```

```python
def extract_rows(payload):
    if isinstance(payload, list): return payload
    for key in ('records', 'predictions', 'items'):
        if isinstance(payload.get(key), list): return payload[key]
    raise ValueError('unsupported prediction JSON schema')
summary = {}
for name, path in paths.items():
    rows = extract_rows(json.loads(path.read_text(encoding='utf-8')))
    summary[name] = {'records': len(rows), 'fields': sorted(rows[0]) if rows else []}
summary
```
