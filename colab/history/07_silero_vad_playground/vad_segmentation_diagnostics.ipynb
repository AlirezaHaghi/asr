# VAD segmentation diagnostics / بررسی قطعه‌بندی
Retrospective reconstruction for examining duration and coverage; results require a rerun with a local WAV file.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import soundfile as sf
AUDIO_PATH = Path('sample.wav')
SEGMENTS = []  # fill with {'start_s': ..., 'end_s': ...} from a rerun
print('ready:', AUDIO_PATH.exists())
```

```python
if AUDIO_PATH.exists():
    audio, sr = sf.read(AUDIO_PATH, dtype='float32', always_2d=True)
    mono = audio.mean(axis=1)
    t = np.arange(len(mono)) / sr
    fig, ax = plt.subplots(figsize=(14, 3))
    ax.plot(t, mono, linewidth=0.35)
    for span in SEGMENTS:
        ax.axvspan(span['start_s'], span['end_s'], color='tab:green', alpha=0.25)
    ax.set(xlabel='seconds', ylabel='amplitude', title='Observed VAD regions')
    plt.show()
else:
    print('Place sample.wav beside the notebook before running diagnostics.')
```

```python
durations = np.array([s['end_s'] - s['start_s'] for s in SEGMENTS], dtype=float)
{'count': int(durations.size), 'mean_s': float(durations.mean()) if durations.size else None, 'p90_s': float(np.quantile(durations, .9)) if durations.size else None}
```
