# Gate threshold sweep / بررسی آستانه حذف نویز

A deterministic synthetic experiment explores how threshold margin can suppress noise and quiet signal alike.

```python
import numpy as np
rng = np.random.default_rng(23)
rate=16000; n=rate*3; t=np.arange(n)/rate
active=((t>.5)&(t<1.4))|((t>1.8)&(t<2.7))
clean=.12*np.sin(2*np.pi*220*t)*active
noise=.025*rng.standard_normal(n)
mixture=clean+noise
noise_rms=np.sqrt(np.mean(mixture[~active]**2))
```

For each margin, use an oracle activity mask only to quantify the trade-off in this research toy example.

```python
rows=[]
for margin_db in [0,3,6,9,12]:
    threshold=noise_rms*10**(margin_db/20)
    gain=np.where(np.abs(mixture)>=threshold,1.0,.1)
    y=mixture*gain
    residual=np.sqrt(np.mean(y[~active]**2)); distortion=np.sqrt(np.mean((y[active]-clean[active])**2))
    rows.append({'margin_db':margin_db,'threshold':float(threshold),'residual_noise_rms':float(residual),
                 'active_rmse':float(distortion),'retained_fraction':float(np.mean(gain==1))})
rows
```
