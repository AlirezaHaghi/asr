This notebook measures simple transforms on synthetic radio-like audio, not ASR accuracy.

```python
import numpy as np
from scipy.signal import lfilter
rng = np.random.default_rng(7)
rate = 16000; t = np.arange(2*rate)/rate
carrier = .12*np.sin(2*np.pi*180*t)*(1+.5*np.sin(2*np.pi*3*t))
noise = .035*rng.standard_normal(len(t))
x = carrier + noise + .02
peak = np.max(np.abs(x)); normalized = x*(10**(-3/20)/peak)
pre = lfilter([1, -.97], [1], normalized)
len(x)
```

Compare scale, DC, and high-frequency difference energy. These diagnostics do not imply lower WER.

```python
def stats(a):
    return {'peak': float(np.max(np.abs(a))), 'rms': float(np.sqrt(np.mean(a*a))),
            'dc': float(np.mean(a)), 'difference_rms': float(np.sqrt(np.mean(np.diff(a)**2)))}
print('raw', stats(x))
print('normalized', stats(normalized))
print('preemphasized', stats(pre))
```
