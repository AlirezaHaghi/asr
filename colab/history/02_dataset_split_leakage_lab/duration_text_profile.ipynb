# Duration and text profile / پروفایل زمان و متن
Retrospective reconstruction from the final notebook/report; results require a rerun.
A bounded descriptive profile; values must be regenerated.

```python
import numpy as np
from datasets import load_dataset
ds = load_dataset('Jzuluaga/atco2_corpus_1h', split='test')
limit = min(100, len(ds))
durations, word_counts = [], []
for row in ds.select(range(limit)):
    a = row['audio']; text = row.get('text', row.get('transcription', ''))
    durations.append(len(a['array']) / a['sampling_rate'])
    word_counts.append(len(text.split()))
```

Quantiles expose long-tail clips that may interact with 30-second chunking.

```python
def summary(values):
    return {'count': len(values), 'min': float(np.min(values)), 'median': float(np.median(values)),
            'p90': float(np.quantile(values, .9)), 'max': float(np.max(values)), 'mean': float(np.mean(values))}
print('duration_s', summary(durations))
print('word_count', summary(word_counts))
print('correlation', float(np.corrcoef(durations, word_counts)[0, 1]))
```
