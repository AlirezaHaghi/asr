Loads a small slice of the project dataset. Network access may be required.

```python
from datasets import load_dataset
DATASET_ID = 'Jzuluaga/atco2_corpus_1h'
dataset = load_dataset(DATASET_ID, split='test')
print(dataset)
print(dataset.features)
```

Avoid printing raw audio arrays; summarize shape, rate, and transcript instead.

```python
for i in range(min(3, len(dataset))):
    row = dataset[i]
    audio = row['audio']
    text_key = 'text' if 'text' in row else 'transcription'
    print({'index': i, 'id': row.get('id'), 'text': row.get(text_key, ''),
           'sampling_rate': audio['sampling_rate'], 'samples': len(audio['array']),
           'duration_s': round(len(audio['array']) / audio['sampling_rate'], 3)})
```
