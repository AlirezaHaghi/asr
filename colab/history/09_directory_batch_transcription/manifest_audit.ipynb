# Manifest audit / ممیزی فهرست صوت
Retrospective reconstruction created from the final notebook/report. It is not claimed to be an original timestamped development artifact; generated metrics require rerunning. Directory-WAV handling is a recreated exploration.

```python
import json
from pathlib import Path
from collections import Counter
MANIFEST = Path('manifest.jsonl')
records = [json.loads(line) for line in MANIFEST.read_text(encoding='utf-8').splitlines() if line.strip()] if MANIFEST.exists() else []
print({'exists': MANIFEST.exists(), 'records': len(records)})
```

```python
hash_counts = Counter(r.get('sha256') for r in records if r.get('sha256'))
duplicates = {digest: count for digest, count in hash_counts.items() if count > 1}
sample_rates = Counter(r.get('sample_rate') for r in records if r.get('valid'))
channels = Counter(r.get('channels') for r in records if r.get('valid'))
{'valid': sum(bool(r.get('valid')) for r in records), 'referenced': sum(r.get('reference') is not None for r in records), 'sample_rates': sample_rates, 'channels': channels, 'duplicate_hashes': duplicates}
```

```python
durations = sorted(float(r['duration_s']) for r in records if r.get('duration_s') is not None)
{'audio_hours': sum(durations)/3600, 'shortest_s': durations[0] if durations else None, 'longest_s': durations[-1] if durations else None}
```
