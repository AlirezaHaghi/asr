# Batch results analysis / تحلیل اجرای پوشه‌ای
Retrospective reconstruction created from the final notebook/report. It is not claimed to be an original timestamped development artifact; generated metrics require rerunning.

```python
import json
from pathlib import Path
from collections import Counter
RESULTS = Path('predictions.jsonl')
raw = [json.loads(line) for line in RESULTS.read_text(encoding='utf-8').splitlines() if line.strip()] if RESULTS.exists() else []
latest = {r['id']: r for r in raw}
records = list(latest.values())
Counter(r.get('status') for r in records)
```

```python
ok = [r for r in records if r.get('status') == 'ok']
audio_s = sum(float(r.get('audio_duration_s') or 0) for r in ok)
inference_s = sum(float(r.get('inference_s') or 0) for r in ok)
{'unique_items': len(records), 'successful': len(ok), 'audio_hours': audio_s/3600, 'weighted_rtf': inference_s/audio_s if audio_s else None}
```

```python
failures = [{'id': r.get('id'), 'error': r.get('error')} for r in records if r.get('status') != 'ok']
failures[:20]
```
