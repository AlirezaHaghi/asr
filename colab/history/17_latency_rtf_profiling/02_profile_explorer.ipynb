# Profile explorer / بررسی پروفایل زمان
Retrospective reconstruction; this is not an original timestamped development artifact and results require rerunning. Set `PROFILE_JSON` to a generated profile.

```python
import json
import os
import statistics
from pathlib import Path
path = Path(os.environ['PROFILE_JSON']) if os.environ.get('PROFILE_JSON') else None
payload = json.loads(path.read_text(encoding='utf-8')) if path else {'records': []}
ok = [row for row in payload.get('records', []) if row.get('status') == 'ok' and row.get('rtf') is not None]
{'selected': str(path) if path else None, 'config': payload.get('config'), 'successful': len(ok)}
```

```python
rtf = [float(row['rtf']) for row in ok]
{'mean_rtf': statistics.fmean(rtf) if rtf else None, 'median_rtf': statistics.median(rtf) if rtf else None}
```
