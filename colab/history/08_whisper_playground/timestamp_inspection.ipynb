# Timestamp inspection / بررسی زمان‌بندی
Retrospective reconstruction created from the final notebook/report. It is not claimed to be an original timestamped development artifact; generated metrics require rerunning.

```python
import json
from pathlib import Path
RESULT_PATH = Path('transcript.json')
payload = json.loads(RESULT_PATH.read_text(encoding='utf-8')) if RESULT_PATH.exists() else {'chunks': []}
chunks = payload.get('chunks', [])
print({'result_exists': RESULT_PATH.exists(), 'timestamp_records': len(chunks)})
```

```python
rows = []
for item in chunks:
    start, end = item.get('timestamp', (None, None))
    rows.append({'text': item.get('text','').strip(), 'start_s': start, 'end_s': end, 'duration_s': (end-start) if start is not None and end is not None else None})
rows[:20]
```

```python
overlaps = []
for left, right in zip(rows, rows[1:]):
    if left['end_s'] is not None and right['start_s'] is not None and right['start_s'] < left['end_s']:
        overlaps.append((left['text'], right['text'], left['end_s'] - right['start_s']))
{'overlap_count': len(overlaps), 'examples': overlaps[:5]}
```
