# Single-file Whisper walkthrough / راهنمای یک فایل
Retrospective reconstruction created from the final notebook/report. It is not claimed to be an original timestamped development artifact; generated metrics require rerunning.

```python
from pathlib import Path
import soundfile as sf
AUDIO_PATH = Path('sample.wav')
MODEL_ID = 'jacktol/whisper-large-v3-finetuned-for-ATC'
if AUDIO_PATH.exists():
    info = sf.info(AUDIO_PATH)
    print({'duration_s': info.duration, 'sample_rate': info.samplerate, 'channels': info.channels})
else:
    print('Place sample.wav beside this notebook.')
```

## Inference / استنتاج
The guarded cell downloads the selected model only when a local file exists.

```python
result = None
if AUDIO_PATH.exists():
    import torch
    from transformers import pipeline
    asr = pipeline('automatic-speech-recognition', model=MODEL_ID, device=0 if torch.cuda.is_available() else -1, torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32, chunk_length_s=30)
    result = asr(str(AUDIO_PATH), generate_kwargs={'language':'english','task':'transcribe','temperature':0.0,'num_beams':5}, return_timestamps='word')
result
```
