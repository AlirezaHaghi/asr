# Decoding ablation analysis / تحلیل آزمایش beam
Retrospective reconstruction created from the final notebook/report. It is not claimed to be an original timestamped development artifact; generated metrics require rerunning.

```python
import json
from pathlib import Path
from collections import defaultdict
PATH = Path('observations.jsonl')
raw = [json.loads(line) for line in PATH.read_text(encoding='utf-8').splitlines() if line.strip()] if PATH.exists() else []
rows = list({(r['run_id'], r['id']): r for r in raw}.values())
print({'exists': PATH.exists(), 'latest_rows': len(rows)})
```

```python
groups = defaultdict(list)
for row in rows: groups[row['run_id']].append(row)
table = []
for run_id, items in sorted(groups.items()):
    ok = [r for r in items if r.get('status') == 'ok']
    audio = sum(float(r.get('audio_duration_s') or 0) for r in ok)
    elapsed = sum(float(r.get('inference_s') or 0) for r in ok)
    wers = [float(r['utterance_wer']) for r in ok if r.get('utterance_wer') is not None]
    table.append({'run_id':run_id, 'ok':len(ok), 'failed':len(items)-len(ok), 'mean_utterance_wer':sum(wers)/len(wers) if wers else None, 'weighted_rtf':elapsed/audio if audio else None})
table
```

```python
if table:
    import matplotlib.pyplot as plt
    labels = [r['run_id'] for r in table if r['mean_utterance_wer'] is not None]
    values = [r['mean_utterance_wer'] for r in table if r['mean_utterance_wer'] is not None]
    plt.bar(labels, values); plt.ylabel('observed mean utterance WER'); plt.xticks(rotation=30); plt.show()
else:
    print('Run the ablation before plotting.')
```
