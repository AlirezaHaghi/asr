# ATC normalizer rulebook / دفترچه قواعد

The cells reproduce the final comparison-normalization rules on transparent examples.

```python
import re
digits=dict(zip('0123456789','zero one two three four five six seven eight nine'.split()))
def normalize(text):
    t=re.sub(r'[,\.!?;:"()]',' ',text.lower().strip())
    for p,r in [(r'\bniner\b','nine'),(r'\balfa\b','alpha'),(r'\btree\b','three'),(r'\bfife\b','five'),
                (r'\bryan\s+air\b','ryanair'),(r'\beuro\s+wings\b','eurowings'),(r'\bspeed\s+bird\b','speedbird')]:
        t=re.sub(p,r,t)
    words=lambda s:' '.join(digits[d] for d in s)
    t=re.sub(r'\bfl\s*(\d{2,3})\b',lambda m:'flight level '+words(m.group(1)),t)
    def runway(m):
        side={'l':'left','r':'right','c':'center'}.get((m.group(2) or '').lower(),'')
        return ' '.join(x for x in ('runway',words(m.group(1)),side) if x)
    t=re.sub(r'\brwy\s*(\d{1,2})([lrc]?)\b',runway,t)
    t=re.sub(r'\b(\d+)\b',lambda m:words(m.group(1)),t)
    return re.sub(r'\s+',' ',t).strip()
```

Idempotence matters: applying the normalizer twice should not keep changing text.

```python
examples=['Descend FL180.', 'Cleared RWY28L!', 'Ryan Air niner, squawk 4567', 'Heading 240']
rows=[{'raw':x,'normalized':normalize(x),'idempotent':normalize(normalize(x))==normalize(x)} for x in examples]
rows
```
