# WER normalization study / مطالعه نرمال‌سازی WER

A small illustrative pair set shows why formatting-equivalent ATC text needs controlled normalization.

```python
import re
from jiwer import wer
digits=dict(zip('0123456789','zero one two three four five six seven eight nine'.split()))
def basic(text):
    t=re.sub(r'[,\.!?;:"()]',' ',text.lower())
    t=re.sub(r'\bniner\b','nine',t)
    words=lambda s:' '.join(digits[d] for d in s)
    t=re.sub(r'\bfl\s*(\d{2,3})\b',lambda m:'flight level '+words(m.group(1)),t)
    t=re.sub(r'\b(\d+)\b',lambda m:words(m.group(1)),t)
    return re.sub(r'\s+',' ',t).strip()
refs=['descend flight level one eight zero','squawk four five six seven','nine thousand feet']
hyps=['Descend FL180.','squawk 4567','niner thousand feet']
```

The example is explanatory only; project metrics must be recomputed from the full prediction files.

```python
raw=wer([x.lower() for x in refs],[x.lower() for x in hyps])
normalized=wer([basic(x) for x in refs],[basic(x) for x in hyps])
print({'example_pairs':len(refs),'raw_wer':raw,'normalized_wer':normalized,'delta':raw-normalized})
list(zip(refs,hyps,map(basic,refs),map(basic,hyps)))
```
