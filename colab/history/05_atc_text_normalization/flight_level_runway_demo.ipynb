# Flight level and runway display / نمایش FL و RWY
Retrospective reconstruction created from the final notebook/report; results require rerunning.
Explore the bounded spoken-number grammar and runway side suffixes.

```python
import re
mapping=dict(zip('zero one two three four five six seven eight nine'.split(),'0123456789'))
mapping['niner']='9'; p='|'.join(mapping)
digits=lambda s:''.join(mapping[w.lower()] for w in s.split())
def format_display(text):
    text=re.sub(rf'\bflight level\s+((?:{p})(?:\s+(?:{p}))*)',lambda m:'FL'+digits(m.group(1)),text,flags=re.I)
    def rw(m):
        side={'left':'L','right':'R','center':'C'}.get((m.group(2) or '').lower(),'')
        return 'RWY'+digits(m.group(1))+side
    return re.sub(rf'\brunway\s+((?:{p})(?:\s+(?:{p}))*)\s*(left|right|center)?',rw,text,flags=re.I)
```

These are formatting cases, not measured ASR predictions.

```python
cases=['climb flight level three five zero','descend flight level eight zero',
       'cleared runway two eight left','hold short runway zero nine center','approach runway one right']
[{'spoken':x,'display':format_display(x)} for x in cases]
```
