# Frequency display / نمایش فرکانس
Retrospective reconstruction created from the final notebook/report; results require rerunning.
Compress explicit spoken decimal frequencies while leaving ambiguous phrases unchanged.

```python
import re
words=dict(zip('zero one two three four five six seven eight nine'.split(),'0123456789')); words['niner']='9'
p='|'.join(words); digits=lambda s:''.join(words[w.lower()] for w in s.split())
def frequency(text):
    pattern=rf'((?:{p})(?:\s+(?:{p}))*)\s+decimal\s+((?:{p})(?:\s+(?:{p}))*)'
    return re.sub(pattern,lambda m:digits(m.group(1))+'.'+digits(m.group(2)),text,flags=re.I)
```

The formatter requires the word `decimal`; bare number sequences stay untouched by design.

```python
cases=['contact one two seven decimal four','monitor one one eight decimal zero five',
       'squawk four five six seven','heading two four zero','contact niner decimal five']
[{'spoken':x,'display':frequency(x),'changed':frequency(x)!=x} for x in cases]
```
