# Alignment error lab / آزمایش هم‌ترازی خطا
Retrospective reconstruction created from the final notebook/report. It is not claimed to be an original timestamped development artifact; generated metrics require rerunning.

```python
import json
from pathlib import Path
from jiwer import process_words
PATH = Path('architecture_observations.jsonl')
rows = [json.loads(line) for line in PATH.read_text(encoding='utf-8').splitlines() if line.strip()] if PATH.exists() else []
latest = list({(r['run_id'],r['id']):r for r in rows}.values())
print({'exists':PATH.exists(), 'latest_rows':len(latest)})
```

```python
examples = []
for row in latest:
    ref, hyp = row.get('reference_normalized'), row.get('hypothesis_normalized')
    if row.get('status') != 'ok' or ref is None: continue
    aligned = process_words(ref, hyp)
    if aligned.substitutions + aligned.deletions + aligned.insertions:
        examples.append({'run_id':row['run_id'],'id':row['id'],'S':aligned.substitutions,'D':aligned.deletions,'I':aligned.insertions,'reference':ref,'hypothesis':hyp,'alignment':aligned.alignments[0]})
examples[:10]
```

```python
from collections import Counter
error_totals = Counter()
for row in examples:
    error_totals.update({'S':row['S'],'D':row['D'],'I':row['I']})
error_totals
```
