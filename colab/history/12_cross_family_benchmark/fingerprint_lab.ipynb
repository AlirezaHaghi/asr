```python
import hashlib, json
def fingerprint(value):
    raw = json.dumps(value, sort_keys=True, separators=(',', ':'), allow_nan=False).encode('utf-8')
    return hashlib.sha256(raw).hexdigest()
left = {'model': 'user/model', 'generate_kwargs': {'num_beams': 1}}
right = {'generate_kwargs': {'num_beams': 1}, 'model': 'user/model'}
fingerprint(left), fingerprint(right)
```

```python
assert fingerprint(left) == fingerprint(right)
changed = {**left, 'generate_kwargs': {'num_beams': 5}}
assert fingerprint(left) != fingerprint(changed)
'canonical key order is stable; beam changes provenance'
```
