# Cache manifest audit / بررسی manifest کش
Retrospective reconstruction; this is not an original timestamped development artifact and results require rerunning. Set `CACHE_MANIFEST` to inspect a generated manifest.

```python
import json, os
from collections import Counter
from pathlib import Path
path = Path(os.environ['CACHE_MANIFEST']) if os.environ.get('CACHE_MANIFEST') else None
manifest = json.loads(path.read_text(encoding='utf-8')) if path else {'entries': []}
entries = manifest.get('entries', [])
{'selected': str(path) if path else None, 'entries': len(entries)}
```

```python
fingerprints = Counter(item.get('config_sha256') for item in entries)
{'duplicate_fingerprints': {key: count for key, count in fingerprints.items() if key and count > 1}, 'missing_checksums': sum(not item.get('predictions_sha256') for item in entries)}
```
