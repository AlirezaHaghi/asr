# Runtime inventory / فهرست محیط
Retrospective reconstruction from the final notebook/report; results require a rerun.
A no-download inspection of the active Python runtime.

```python
import json
import os
import platform
import shutil
import sys
from pathlib import Path
disk = shutil.disk_usage(Path.cwd())
inventory = {
    'python': sys.version, 'platform': platform.platform(),
    'machine': platform.machine(), 'cpu_count': os.cpu_count(),
    'cwd': str(Path.cwd()), 'disk_free_gib': round(disk.free / 2**30, 2),
}
print(json.dumps(inventory, indent=2))
```

The values above are observations from the current rerun, not historical benchmark evidence.
