Checks CUDA and numeric dtypes without loading an ASR model.

```python
try:
    import torch
except ImportError:
    torch = None
status = {'torch_installed': torch is not None}
if torch is not None:
    status.update(torch_version=torch.__version__, cuda_available=torch.cuda.is_available(), device_count=torch.cuda.device_count())
    status['devices'] = [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())]
status
```

A small matrix multiplication is only a smoke test; it does not estimate full-model memory.

```python
if torch is None:
    print('Install requirements.txt to run the tensor probe')
else:
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    dtype = torch.float16 if device == 'cuda' else torch.float32
    x = torch.randn((256, 256), device=device, dtype=dtype)
    y = x @ x.T
    print({'device': device, 'dtype': str(dtype), 'shape': tuple(y.shape), 'finite': bool(torch.isfinite(y).all())})
```
