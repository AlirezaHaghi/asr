```python
import glob, json, os
from pathlib import Path
pattern = os.environ.get('PREDICTION_GLOB', 'no_predictions_selected_*.json')
files = [Path(name) for name in sorted(glob.glob(pattern))]
files
```

```python
def rows(payload):
    if isinstance(payload, list): return payload
    for key in ('records', 'predictions', 'items'):
        if isinstance(payload.get(key), list): return payload[key]
    raise ValueError('unsupported schema')
inventory = [{'path': str(path), 'records': len(rows(json.loads(path.read_text(encoding='utf-8'))))} for path in files]
inventory
```
