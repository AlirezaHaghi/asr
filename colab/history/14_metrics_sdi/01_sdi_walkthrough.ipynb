```python
def edit_distance(ref, hyp):
    previous = list(range(len(hyp)+1))
    for i, left in enumerate(ref, 1):
        current = [i]
        for j, right in enumerate(hyp, 1):
            current.append(previous[j-1] if left == right else 1+min(previous[j-1], previous[j], current[-1]))
        previous = current
    return previous[-1]
examples = [('alpha bravo'.split(), 'alpha extra bravo'.split()), ('runway two'.split(), 'runway three'.split())]
[edit_distance(ref, hyp) for ref, hyp in examples]
```
