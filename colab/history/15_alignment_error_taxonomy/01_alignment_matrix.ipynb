# Alignment matrix / ماتریس هم‌ترازی

```python
reference = 'contact tower'.split()
hypothesis = 'contact the tower'.split()
matrix = [[0]*(len(hypothesis)+1) for _ in range(len(reference)+1)]
for i in range(len(reference)+1): matrix[i][0] = i
for j in range(len(hypothesis)+1): matrix[0][j] = j
for i in range(1, len(reference)+1):
    for j in range(1, len(hypothesis)+1):
        matrix[i][j] = matrix[i-1][j-1] if reference[i-1] == hypothesis[j-1] else 1+min(matrix[i-1][j-1], matrix[i-1][j], matrix[i][j-1])
matrix
```
